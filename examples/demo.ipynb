# dms-convert demo

Parse messy coordinate strings and print them in three formats, then run the parser over the real landmark table in `examples/landmarks.csv`. Offline, no extra dependencies.

```python
# Setup: install the package when it is missing (e.g. a fresh Colab runtime)
try:
    import dms_convert  # noqa: F401
except ImportError:
    import subprocess
    import sys

    subprocess.check_call([sys.executable, "-m", "pip", "install", "-q",
                           "git+https://github.com/AmmarYasser455/dms-convert.git"])

from dms_convert import parse_coordinate, format_coordinate

lat, lon = parse_coordinate("40°26'46\"N 79°58'56\"W")  # (lat, lon) order
print(round(lat, 4), round(lon, 4))
print(format_coordinate(lat, lon, "ddm"))
print(format_coordinate(lat, lon, "dms", precision=0))
```

```python
import csv
import io
import os

# six landmarks (Wikidata, CC0), each written in a different text style
path = next((p for p in ("landmarks.csv", "examples/landmarks.csv") if os.path.exists(p)), None)
if path:
    text = open(path, encoding="utf-8", newline="").read()
else:  # e.g. on Colab, where the repository is not checked out
    from urllib.request import urlopen
    url = "https://raw.githubusercontent.com/AmmarYasser455/dms-convert/main/examples/landmarks.csv"
    text = urlopen(url).read().decode("utf-8")
for row in csv.DictReader(io.StringIO(text)):
    lat, lon = parse_coordinate(row["coordinate"])
    print(f"{row['name']}: {lat:.4f}, {lon:.4f}")
```
