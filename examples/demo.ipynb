# dms-convert demo

Parse messy coordinate strings and print them in three formats. Offline, no extra dependencies.

```python
from dms_convert import parse_coordinate, format_coordinate

lat, lon = parse_coordinate("40°26'46\"N 79°58'56\"W")  # (lat, lon) order
print(round(lat, 4), round(lon, 4))
print(format_coordinate(lat, lon, "ddm"))
print(format_coordinate(lat, lon, "dms", precision=0))
```
